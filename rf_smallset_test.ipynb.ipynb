```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Load a sample of the data
df = pd.read_csv('wildfire_data_with_clusters.csv', nrows=10000)
print(f"Data loaded: {df.shape[0]:,} rows")
```

```python
# Features and Target
weather_cols = [
    'pr','rmax','rmin','sph','srad','tmmn','tmmx','vs',
    'bi','fm100','fm1000','erc','etr','pet','vpd'
]
geo_cols = ['latitude', 'longitude']
cluster_col = ['cluster_id']
feature_cols = geo_cols + weather_cols + cluster_col

X = df[feature_cols]
y = df['Wildfire_binary']
```

```python
# Stratified Shuffle Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print(f"Train size: {len(X_train)}, Test size: {len(X_test)}")
```

```python
# Calculate Sample Weights
fire_weight = (y_train == 0).sum() / max(y_train.sum(), 1)
sample_weight = np.where(y_train==1, fire_weight, 1)
print(f"Fire weight: {fire_weight:.2f}")

# Train Model
rf_model = RandomForestClassifier(n_estimators=200, max_depth=12, min_samples_leaf=5, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train, sample_weight=sample_weight)
print("Training complete.")
```

```python
# Evaluate
y_pred = rf_model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")
print("Classification Report:")
print(classification_report(y_test, y_pred))

print("Verification successful!")
```
